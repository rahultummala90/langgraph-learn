#### Pydantic Basics


```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):
    name: str
    age: int
    city: str

person = Person(name="Rahul", age=36, city="Atlanta")
print(person)

personError = Person(name="Rahul", age=36, city=12)
print(personError)
```

```python
type(person)
```

```python
from dataclasses import dataclass

@dataclass
class Person1():
    name: str
    age: int
    city: str

person1 = Person1(name="Rahul", age=36, city=12)
print(person1)
```

#### Model with optional fields

```python
from typing import Optional

class Employee(BaseModel):
    id: int
    name: str
    department: str
    salary: Optional[float] = None
    is_active: Optional[bool] = True
```

```python
employee1 = Employee(id=1, name="Rahul", department="IT")
print(employee1)
```

```python
employee2 = Employee(id=2, name="Alice", department="HR", salary=75000, is_active=False)
print(employee2)
```

```python
from pydantic import BaseModel
from typing import List

class Classroom(BaseModel):
    room_number: str
    students: List[str]
    capacity: int
```

```python
classroom = Classroom(room_number="101A", students=["Alice", "Bob", "Charlie"], capacity=30)
print(classroom)
```

```python
try:
    invalid_classroom = Classroom(room_number="101A", students=["Alice", "Bob", "Charlie"], capacity="thirty")
except Exception as e:
    print("Error creating classroom:", e)
```

#### Nested Models

```python
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    zip_code: int

class Customer(BaseModel):
    name: str
    email: str
    address: Address

customer = Customer(
    name="John Doe",
    email="john.doe@example.com",
    address=Address(
        street="123 Main St",
        city="Anytown",
        zip_code="12345"
    )
)
print(customer)
```

#### Pydantic fields - customization and constraints

```python
from pydantic import BaseModel,Field

class Item(BaseModel):
    name: str=Field(min_length=2, max_length=50)
    price: float=Field(gt=0,le=1000)
    quantity: int=Field(ge=1)

item = Item(name="Laptop", price=999.99, quantity=1)
print(item)

# itemError = Item(name="L", price=-10, quantity=0)
# print(itemError)
```

```python
from pydantic import BaseModel,Field

class User(BaseModel):
    username: str=Field(..., description="The unique username for the user")
    age: int=Field(default=18, description="The age of the user")
    email: str=Field(default_factory=lambda: "user@example.com", description="The email address of the user")

user1 = User(username="john_doe")
print(user1)

user2 = User(username="jane_doe", age=25, email="jane@example.com")
print(user2)
```
